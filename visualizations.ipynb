# Runtime Comparisons

```python
import matplotlib.pyplot as plt
import numpy as np

# Data
vector_sizes = ['2^20', '2^24', '2^30']
asm_times = [0.0019, 0.0267, 1.48665]
c_times = [0.00385, 0.05925, 3.73525]

x = np.arange(len(vector_sizes))
width = 0.35  

fig, ax = plt.subplots(figsize=(10, 6))
rects1 = ax.bar(x - width/2, asm_times, width, label='Assembly', color='skyblue')
rects2 = ax.bar(x + width/2, c_times, width, label='C', color='orange')

ax.set_ylabel('Average Time (seconds)')
ax.set_xlabel('Vector Size')
ax.set_title('Performance Comparison: Assembly vs C')
ax.set_xticks(x)
ax.set_xticklabels(vector_sizes)
ax.legend()

ax.bar_label(rects1, padding=3, fmt='%.6f')
ax.bar_label(rects2, padding=3, fmt='%.6f')

fig.tight_layout()

plt.show()
```
